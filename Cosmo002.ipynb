```python
import numpy as np

# -------------------------------
# Step 1: Entropic Inflation Model
# -------------------------------
def entropic_inflation(delta, kappa, steps=100):
    """
    Simulate expansion driven by entropic gradient δ - κ.
    Returns expansion profile and inflation metrics.
    """
    gradient = delta - kappa
    expansion = np.zeros(steps)
    entropy = np.zeros(steps)

    for t in range(steps):
        expansion[t] = np.exp(gradient * t / steps)  # exponential growth
        entropy[t] = np.log1p(expansion[t]) / (1 + t)  # entropy damping

    inflation_rate = expansion[10] / expansion[0]
    exit_slope = expansion[-1] - expansion[-2]
    return expansion, entropy, inflation_rate, exit_slope

# -------------------------------
# Step 2: Sweep δ and κ
# -------------------------------
delta_values = np.linspace(0.6, 0.8, 10)
kappa_values = np.linspace(0.6, 0.8, 10)
results = []

print("🪐 Act XXXIV: Entropic Inflation")
for delta in delta_values:
    for kappa in kappa_values:
        expansion, entropy, inflation_rate, exit_slope = entropic_inflation(delta, kappa)
        label = "✓" if inflation_rate > 2.0 else "–"
        print(f"{label} δ = {delta:.4f}, κ = {kappa:.4f}, δ−κ = {delta - kappa:+.4f} | Inflation Rate = {inflation_rate:.2f}, Exit Slope = {exit_slope:.4f}")
```
