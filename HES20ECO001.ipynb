```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
M = 10.0         # mass
alpha = 0.1      # gravitational coupling
beta = 0.05      # entropic scaling
gamma = 0.2      # entropy coefficient
R_c = 1.0        # critical collapse scale
timesteps = 200

def entropy_term(R):
    return gamma * np.log(1 + R/R_c)

def collapse_step(R):
    return R - alpha * M / (R**2) + beta * entropy_term(R)

# Simulation
R_vals = [10.0]  # initial radius
for t in range(timesteps):
    R_next = collapse_step(R_vals[-1])
    R_vals.append(max(R_next, R_c))  # enforce non-singularity

# Plot
plt.plot(R_vals, label="Collapse trajectory")
plt.axhline(R_c, color='r', linestyle='--', label="Critical scale")
plt.xlabel("Timestep")
plt.ylabel("Radius")
plt.legend()
plt.title("ECO001: Entropic Collapse Simulation")
plt.show()
```
