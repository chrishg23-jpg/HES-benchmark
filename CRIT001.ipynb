```python
import numpy as np
import matplotlib.pyplot as plt
import os

# ============================================================
# 1. PARAMETERS — CHANGE THESE AS YOU LIKE
# ============================================================

N = 256                 # lattice size (256x256)
timesteps = 5000        # total evolution steps
save_every = 200        # save a frame every N steps
lambda_coupling = 1.0   # CHANGE THIS for each run
save_folder = f"/content/lambda_{lambda_coupling}/"

os.makedirs(save_folder, exist_ok=True)

# ============================================================
# 2. INITIAL FIELD — RANDOM NOISE
# ============================================================

field = np.random.randn(N, N) * 0.1

# Precompute Fourier grid
kx = np.fft.fftfreq(N) * N
ky = np.fft.fftfreq(N) * N
KX, KY = np.meshgrid(kx, ky)
K2 = KX**2 + KY**2

# ============================================================
# 3. FILTERS (Holographic Sequestration)
# ============================================================

# High-frequency damping (Σ)
# You used something like a soft cutoff around mode 100
k_cut = 100
sigma = 15
sequestration_filter = np.exp(-((np.sqrt(K2) - k_cut)/sigma)**2)

# Low-frequency stabilizer (prevents drift)
low_k_filter = np.exp(-K2 * 0.0005)

# Combined filter
SF_filter = sequestration_filter * low_k_filter

# ============================================================
# 4. UPDATE LOOP (Spectral + Filter)
# ============================================================

def evolve_step(field, lam):
    # Fourier transform
    F = np.fft.fft2(field)

    # Apply holographic sequestration filter
    F_filtered = F * SF_filter

    # Back to real space
    filtered = np.fft.ifft2(F_filtered).real

    # Nonlinear update (your λ coupling)
    updated = field + lam * (filtered - field)

    return updated

# ============================================================
# 5. RUN SIMULATION
# ============================================================

for t in range(timesteps):
    field = evolve_step(field, lambda_coupling)

    if t % save_every == 0:
        np.save(save_folder + f"frame_{t}.npy", field)
        print("Saved frame", t)

print("Done. Frames saved in:", save_folder)
```
