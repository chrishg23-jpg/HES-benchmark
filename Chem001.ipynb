```python
# Chem001_Act01: Orbital Encoding
# Simulate electron configurations using HES curvature wells

import matplotlib.pyplot as plt

# Define orbital structure
orbitals = [
    {"name": "1s", "capacity": 2, "entropy_cost": 1},
    {"name": "2s", "capacity": 2, "entropy_cost": 2},
    {"name": "2p", "capacity": 6, "entropy_cost": 3},
    {"name": "3s", "capacity": 2, "entropy_cost": 4},
    {"name": "3p", "capacity": 6, "entropy_cost": 5},
    {"name": "4s", "capacity": 2, "entropy_cost": 6},
    {"name": "3d", "capacity": 10, "entropy_cost": 7},
    {"name": "4p", "capacity": 6, "entropy_cost": 8}
]

# Simulate electron filling
def fill_orbitals(Z):
    electrons = Z
    config = {}
    total_entropy = 0
    for orb in orbitals:
        fill = min(electrons, orb["capacity"])
        config[orb["name"]] = fill
        total_entropy += fill * orb["entropy_cost"]
        electrons -= fill
        if electrons == 0:
            break
    return config, total_entropy

# Run simulation for Z = 1 to 20
results = []
for Z in range(1, 21):
    config, entropy = fill_orbitals(Z)
    results.append({"Z": Z, "config": config, "entropy": entropy})

# Display results
for r in results:
    print(f"Z = {r['Z']}: {r['config']} | Entropic Cost = {r['entropy']}")

# Plot entropy trend
plt.figure(figsize=(10, 5))
plt.plot([r["Z"] for r in results], [r["entropy"] for r in results], marker='o')
plt.title("Entropic Cost vs Atomic Number (Z)")
plt.xlabel("Atomic Number (Z)")
plt.ylabel("Total Entropic Cost")
plt.grid(True)
plt.show()
```
