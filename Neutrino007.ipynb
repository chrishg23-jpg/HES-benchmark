```python
# 🔬 Level 4 Sweep – Neutrino Hierarchy via δ/κ
import numpy as np

# -------------------------------
# Core Analysis Functions
# -------------------------------
def extract_mass_matrix(A_flavors):
    masses = np.sort(np.array(A_flavors))
    return masses**2

def compute_mass_differences(mass_squared):
    return mass_squared[1] - mass_squared[0], mass_squared[2] - mass_squared[1]

def compute_mixing_angles(A_flavors):
    norms = [a / np.linalg.norm(a) for a in A_flavors]
    overlaps = np.zeros((3, 3))
    for i in range(3):
        for j in range(3):
            overlaps[i, j] = np.abs(np.dot(norms[i], norms[j]))
    return overlaps

def calibrate_mass_units(sim_mass_values, reference_scale_eV):
    sim_spread = np.max(sim_mass_values) - np.min(sim_mass_values)
    scale_factor = reference_scale_eV / sim_spread
    return np.array(sim_mass_values) * scale_factor

# -------------------------------
# Sweep Parameters
# -------------------------------
delta_k_values = [0.01, 0.05, 0.10, 0.15, 0.20]
simulation_results = {}

# -------------------------------
# Simulation Function (Replace with real data if available)
# -------------------------------
def simulate_flavor_amplitudes(delta_k):
    base = 1.00
    spread = 0.12 * delta_k / 0.20  # scale spread with delta_k
    return [base, base + spread/2, base + spread]

def simulate_flavor_vectors():
    return [np.random.rand(100) for _ in range(3)]

# -------------------------------
# Run Sweep
# -------------------------------
print("🚀 Running Level 4 Sweep Across δ/κ...\n")

for dk in delta_k_values:
    print(f"🔹 δ/κ = {dk:.2f}")

    A_flavors = simulate_flavor_amplitudes(dk)
    flavor_vectors = simulate_flavor_vectors()

    mass_squared = extract_mass_matrix(A_flavors)
    delta_m2_21, delta_m2_32 = compute_mass_differences(mass_squared)
    mixing_matrix = compute_mixing_angles(flavor_vectors)
    calibrated_masses = calibrate_mass_units(A_flavors, reference_scale_eV=0.05)

    simulation_results[dk] = (delta_m2_21, delta_m2_32)

    print(f"  Mass Estimates: {np.round(A_flavors, 6)}")
    print(f"  Mass²: {np.round(mass_squared, 6)}")
    print(f"  Δm²_21 = {delta_m2_21:.6f}, Δm²_32 = {delta_m2_32:.6f}")
    print(f"  Calibrated Masses (eV/c²): {np.round(calibrated_masses, 6)}")
    print(f"  Mixing Matrix:\n{np.round(mixing_matrix, 4)}\n")

print("✅ Sweep complete. Codex inscription recommended.\n")

# -------------------------------
# Optional: Match to Real Neutrino Hierarchy
# -------------------------------
def find_physical_match():
    target_solar = 7.5e-5  # eV²
    target_atmospheric = 2.5e-3  # eV²
    real_ratio = target_atmospheric / target_solar  # ≈ 33.3

    print("🔍 Searching for physical match to real neutrino hierarchy...\n")
    for dk in delta_k_values:
        sim_d21, sim_d32 = simulation_results[dk]
        ratio = sim_d32 / sim_d21
        if abs(ratio - real_ratio) < 5:
            print(f"🚨 PHYSICAL MATCH FOUND at δ/κ = {dk:.2f}")
            print(f"   Simulated ratio = {ratio:.2f}, Real ratio ≈ {real_ratio:.2f}")
            print(f"   Δm²_21 = {sim_d21:.6f}, Δm²_32 = {sim_d32:.6f}\n")

find_physical_match()
```
