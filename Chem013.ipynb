```python
# Chem013: Reaction Kinetics via Curvature Barriers and Entropic Flow

import numpy as np
import pandas as pd

# Constants
k_B = 8.617e-5  # Boltzmann constant in eV/K
T = 298         # Temperature in Kelvin

# Define reactions and curvature barrier parameters
reactions = [
    {
        "name": "Combustion of methane",
        "reaction": "CH₄ + 2O₂ → CO₂ + 2H₂O",
        "barrier_height": 1.8,
        "barrier_width": 0.6
    },
    {
        "name": "Formation of ammonia",
        "reaction": "N₂ + 3H₂ → 2NH₃",
        "barrier_height": 2.2,
        "barrier_width": 0.9
    },
    {
        "name": "Dissolution of NaCl",
        "reaction": "NaCl(s) → Na⁺ + Cl⁻",
        "barrier_height": 0.5,
        "barrier_width": 0.3
    },
    {
        "name": "Phase change of H₂O",
        "reaction": "H₂O(s) → H₂O(l)",
        "barrier_height": 0.3,
        "barrier_width": 0.8
    },
    {
        "name": "ATP hydrolysis",
        "reaction": "ATP → ADP + Pi",
        "barrier_height": 1.2,
        "barrier_width": 0.5
    }
]

# Compute kinetics
results = []
for rxn in reactions:
    E_a = rxn["barrier_height"]
    width = rxn["barrier_width"]
    compression = round(E_a / width, 2)

    # Tunneling probability (simplified exponential decay model)
    tunneling = round(np.exp(-compression), 5)

    # Rate constant (Arrhenius-like model)
    rate = round(np.exp(-E_a / (k_B * T)), 2)

    interpretation = f"Barrier height {E_a:.2f} eV, width {width:.2f}, compression {compression:.2f}"

    results.append({
        "Reaction": rxn["reaction"],
        "Activation Energy (eV)": E_a,
        "Tunneling Probability": tunneling,
        "Rate Constant (1/s)": rate,
        "Curvature Interpretation": interpretation
    })

# Create and display dataframe
df = pd.DataFrame(results)
print(df)
```
