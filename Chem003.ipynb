```python
# Chem003: Molecular Geometry from HES curvature-memory fields

# Define molecules with bonded atoms and lone pairs
molecules = {
    "H2O": {"bonded": 2, "lone_pairs": 2},
    "CH4": {"bonded": 4, "lone_pairs": 0},
    "NH3": {"bonded": 3, "lone_pairs": 1}
}

# Estimate geometry based on domain repulsion
def estimate_geometry(mol):
    total_domains = mol["bonded"] + mol["lone_pairs"]
    if total_domains == 2:
        shape = "linear"
        angle = 180
    elif total_domains == 3:
        shape = "trigonal planar" if mol["lone_pairs"] == 0 else "bent"
        angle = 120 if mol["lone_pairs"] == 0 else 104.5
    elif total_domains == 4:
        if mol["lone_pairs"] == 0:
            shape = "tetrahedral"
            angle = 109.5
        elif mol["lone_pairs"] == 1:
            shape = "trigonal pyramidal"
            angle = 107
        elif mol["lone_pairs"] == 2:
            shape = "bent"
            angle = 104.5
        else:
            shape = "distorted tetrahedral"
            angle = 100
    else:
        shape = "unknown"
        angle = None
    return {"shape": shape, "angle": angle}

# Run simulation
print("═" * 60)
print("Chem003: Molecular Geometry Simulation")
print("═" * 60)
for name, mol in molecules.items():
    result = estimate_geometry(mol)
    print(f"{name:5} | Shape: {result['shape']:20} | Bond Angle: {result['angle']}°")
```
