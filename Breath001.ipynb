```python
# BREATH001: Breathing Equation Simulation
import numpy as np
import matplotlib.pyplot as plt

# Parameters
T = 500                  # Total time steps
dt = 1.0                 # Time step size
kappa = 0.05             # Restoring force toward N=100
delta = 5.0              # Chaos leak amplitude
omega = 0.1              # Breathing frequency
N0 = 95.0                # Initial N

# Time array
t = np.arange(0, T, dt)
N = np.zeros_like(t)
N[0] = N0

# Evolution loop
for i in range(1, len(t)):
    dNdt = -kappa * (N[i-1] - 100) + delta * np.sin(omega * t[i-1])
    N[i] = N[i-1] + dt * dNdt

# Plot N(t)
plt.figure(figsize=(8, 4))
plt.plot(t, N, label='N(t)', color='teal')
plt.axhline(100, color='gray', linestyle='--', label='Equilibrium N=100')
plt.title('BREATH001 — Breathing Evolution of N(t)')
plt.xlabel('Time')
plt.ylabel('Dimensional Harmonic N(t)')
plt.legend()
plt.grid(True)
plt.show()

# Print summary stats
print(f"Final N(t=500): {N[-1]:.2f}")
print(f"Mean N(t): {np.mean(N):.2f}")
print(f"Max N(t): {np.max(N):.2f}")
print(f"Min N(t): {np.min(N):.2f}")
print(f"Oscillation Amplitude: ~{(np.max(N) - np.min(N)) / 2:.2f}")
```
