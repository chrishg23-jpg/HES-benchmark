```python
import numpy as np

# -------------------------------
# Step 1: HES Prediction Model
# -------------------------------
def hes_observables(delta, kappa, steps=100):
    """
    Simulate HES predictions for CMB scale, supernova dimming, and structure amplitude.
    """
    curvature = np.zeros(steps)
    structure = np.zeros(steps)
    luminosity = np.zeros(steps)

    for t in range(steps):
        curvature[t] = np.sin(delta * t / steps) + np.cos(kappa * t / steps)
        structure[t] = np.exp(-kappa * t / steps) * 1e-5
        luminosity[t] = 1 + curvature[t] * delta / (1 + t)

    cmb_scale = np.std(curvature[:20])
    sn_dimming = luminosity[-1] - luminosity[0]
    structure_amp = structure[-1]
    return cmb_scale, sn_dimming, structure_amp

# -------------------------------
# Step 2: Sweep δ and κ
# -------------------------------
delta_values = np.linspace(0.6, 0.8, 10)
kappa_values = np.linspace(0.6, 0.8, 10)

print("🪐 Act XXXVIII: Observational Signatures")
for delta in delta_values:
    for kappa in kappa_values:
        cmb, sn, amp = hes_observables(delta, kappa)
        label = "✓" if np.isclose(amp, 1e-5, rtol=0.2) else "–"
        print(f"{label} δ = {delta:.4f}, κ = {kappa:.4f}, δ/κ = {delta/kappa:.3f} | CMB σ = {cmb:.4f}, SN Δ = {sn:.4f}, Structure = {amp:.2e}")
```
