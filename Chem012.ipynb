```python
# Chem012: Thermochemistry via Curvature Accounting and Memory Vector Flow

import numpy as np
import pandas as pd

# Temperature in Kelvin
T = 298

# Define reactions and curvature parameters
reactions = [
    {
        "name": "Combustion of methane",
        "reaction": "CH₄ + 2O₂ → CO₂ + 2H₂O",
        "reactants": {"CH₄": 1.2, "O₂": 2.0},
        "products": {"CO₂": 2.5, "H₂O": 2.2},
        "vector_density": {"CH₄": 0.8, "O₂": 0.9, "CO₂": 1.5, "H₂O": 1.4}
    },
    {
        "name": "Formation of ammonia",
        "reaction": "N₂ + 3H₂ → 2NH₃",
        "reactants": {"N₂": 2.0, "H₂": 1.0},
        "products": {"NH₃": 2.8},
        "vector_density": {"N₂": 0.7, "H₂": 0.6, "NH₃": 1.2}
    },
    {
        "name": "Dissolution of NaCl",
        "reaction": "NaCl(s) → Na⁺(aq) + Cl⁻(aq)",
        "reactants": {"NaCl": 3.0},
        "products": {"Na⁺": 1.5, "Cl⁻": 1.5},
        "vector_density": {"NaCl": 0.5, "Na⁺": 1.0, "Cl⁻": 1.0}
    },
    {
        "name": "Phase change of H₂O",
        "reaction": "H₂O(s) → H₂O(l)",
        "reactants": {"H₂O(s)": 2.5},
        "products": {"H₂O(l)": 2.2},
        "vector_density": {"H₂O(s)": 0.6, "H₂O(l)": 1.0}
    },
    {
        "name": "ATP hydrolysis",
        "reaction": "ATP → ADP + Pi",
        "reactants": {"ATP": 4.0},
        "products": {"ADP": 2.0, "Pi": 1.8},
        "vector_density": {"ATP": 1.0, "ADP": 1.2, "Pi": 1.3}
    }
]

# Compute thermochemical properties
results = []
for rxn in reactions:
    H_react = sum(rxn["reactants"][sp] for sp in rxn["reactants"])
    H_prod = sum(rxn["products"][sp] for sp in rxn["products"])
    ΔH = round(H_prod - H_react, 2)

    S_react = sum(rxn["vector_density"][sp] for sp in rxn["reactants"])
    S_prod = sum(rxn["vector_density"][sp] for sp in rxn["products"])
    ΔS = round(S_prod - S_react, 2)

    ΔG = round(ΔH - T * ΔS / 1000, 2)

    interpretation = f"ΔH from curvature depth ({H_prod:.1f} - {H_react:.1f}), ΔS from vector density ({S_prod:.1f} - {S_react:.1f})"

    results.append({
        "Reaction": rxn["reaction"],
        "ΔH (kJ/mol)": ΔH,
        "ΔS (J/mol·K)": ΔS,
        "ΔG (kJ/mol)": ΔG,
        "Curvature Interpretation": interpretation
    })

# Create and display dataframe
df = pd.DataFrame(results)
print(df)
```
