```python
# 🧬 TEG009: Entropic Caustics
# Tests for scattering, convergence, and echo formation in entropic curvature folds
# Author: Chris & Copilot | Arc VIII: Tensorial Emergence

import numpy as np
import matplotlib.pyplot as plt

# 📐 Curvature Tensor (Folded Field)
def curvature_tensor(x, y, sigma=0.3, epsilon=0.6):
    exp_term = np.exp(-(x**2 + y**2) / (2 * sigma**2))
    fold = np.sin(3 * x) * np.cos(3 * y)  # Caustic fold pattern
    Cxx = (1 - x**2 / sigma**2) * exp_term * (1 + epsilon) + fold
    Cyy = (1 - y**2 / sigma**2) * exp_term * (1 - epsilon) - fold
    Cxy = (-x * y / sigma**2) * exp_term
    return np.array([[Cxx, Cxy], [Cxy, Cyy]])

# 🧲 Tensor Force
def tensor_force(x, y, vx, vy, sigma=0.3, epsilon=0.6):
    C = curvature_tensor(x, y, sigma, epsilon)
    v = np.array([vx, vy])
    f = -np.dot(C, v)
    return f[0], f[1]

# 🔁 Velocity Update
def update_velocity(vx, vy, fx, fy, dt):
    return vx + fx * dt, vy + fy * dt

# 🚀 Simulation Loop
def simulate_caustics(x0, y0, vx0, vy0, steps=5000, dt=0.01, sigma=0.3, epsilon=0.6):
    x, y = x0, y0
    vx, vy = vx0, vy0
    trajectory = []

    for _ in range(steps):
        fx, fy = tensor_force(x, y, vx, vy, sigma, epsilon)
        vx, vy = update_velocity(vx, vy, fx, fy, dt)
        x += vx * dt
        y += vy * dt
        trajectory.append((x, y))

    return np.array(trajectory)

# 📊 Trajectory Plot
def plot_trajectory(trajectory):
    x_vals, y_vals = trajectory[:,0], trajectory[:,1]
    plt.figure(figsize=(8, 8))
    plt.plot(x_vals, y_vals, linewidth=1.5)
    plt.title("TEG009: Entropic Caustics Trajectory")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.grid(True)
    plt.axis('equal')
    plt.show()

# 🧪 Run Caustic Test
x0, y0 = -2.0, 0.0
vx0, vy0 = 1.0, 0.0  # Lightlike launch
steps = 5000
dt = 0.01
sigma = 0.3
epsilon = 0.6

trajectory = simulate_caustics(x0, y0, vx0, vy0, steps, dt, sigma, epsilon)
plot_trajectory(trajectory)
```
