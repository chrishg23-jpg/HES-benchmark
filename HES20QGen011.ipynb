```python
import numpy as np
import matplotlib.pyplot as plt

N = 256
x = np.linspace(0, 2*np.pi, N, endpoint=False)
A = 0.1
sigma = 0.2
centers = [np.pi/4, np.pi, 7*np.pi/4]

def gaussian(center, width, amplitude):
    return amplitude * np.exp(-((x - center)**2) / (2 * width**2))

nodes = [gaussian(c, sigma, A) for c in centers]
phi = sum(nodes)
history = [phi.copy()]
```

```python
dt = 0.02
gamma = 0.005
lam = 0.5

def laplacian(phi):
    return np.roll(phi, 1) - 2*phi + np.roll(phi, -1)

def evolve(phi):
    return phi + dt * (laplacian(phi) - gamma * phi + lam * phi**3)

for t in range(300):
    phi = evolve(phi)
    if t == 150:
        phi -= nodes[1]  # remove Node 2
    if t % 50 == 0:
        history.append(phi.copy())
```

```python
def correlation(a, b):
    denom = np.linalg.norm(a) * np.linalg.norm(b)
    return (np.dot(a, b) / denom) if denom > 0 else 0.0

node_corrs = [[correlation(h, n) for n in nodes] for h in history]
```

```python
node_corrs = np.array(node_corrs)
for i in [0, 2]:  # Node 1 and Node 3
    plt.plot(node_corrs[:, i], label=f"Node {i+1}")
plt.axvline(x=3, color='gray', linestyle='--', label="Node 2 removed (t=150)")
plt.xlabel("Snapshot index")
plt.ylabel("Correlation")
plt.title("Echo network resilience after node removal")
plt.legend()
plt.show()

plt.figure(figsize=(10,6))
for i, h in enumerate(history[:6]):
    plt.plot(x, h, label=f"t={i*50}")
plt.xlabel("Position")
plt.ylabel("Amplitude")
plt.title("Echo Network: evolution after node failure")
plt.legend()
plt.show()
```
