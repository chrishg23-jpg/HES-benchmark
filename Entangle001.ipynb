```python
import numpy as np
import matplotlib.pyplot as plt

# -------------------------------
# Step 1: Initialize Entanglement Grid
# -------------------------------
def initialize_grid(size=32, seed=42):
    np.random.seed(seed)
    return np.random.normal(loc=1.0, scale=0.01, size=size)

# -------------------------------
# Step 2: Entanglement Evolution Rule
# -------------------------------
def evolve_entropy(grid, delta, kappa, steps=100):
    history = [grid.copy()]
    curvature_trace = []

    for t in range(steps):
        safe_max = max(np.max(grid), 1e-6)  # avoid division by zero
        injection = delta * np.sin(np.pi * grid / safe_max)
        safe_grid = np.clip(grid, -10, 10)  # prevent overflow in exp
        damping = kappa * np.exp(-safe_grid)
        update = injection - damping
        grid += update / (1 + t)
        curvature = np.std(update)
        curvature_trace.append(curvature)
        history.append(grid.copy())

    return np.array(history), np.array(curvature_trace)

# -------------------------------
# Step 3: Sweep δ and κ
# -------------------------------
delta_values = np.linspace(0.6, 0.8, 5)
kappa_values = np.linspace(0.6, 0.8, 5)

print("🧠 Arc I: The Entanglement Engine — Calibrated")
for delta in delta_values:
    for kappa in kappa_values:
        grid = initialize_grid()
        history, curvature = evolve_entropy(grid, delta, kappa)
        final_entropy = np.mean(history[-1])
        final_curvature = curvature[-1]
        label = "✓" if final_curvature > 0.01 else "–"
        print(f"{label} δ = {delta:.3f}, κ = {kappa:.3f}, δ/κ = {delta/kappa:.3f} | Final S̄ = {final_entropy:.4f}, Curvature = {final_curvature:.4f}")
```
