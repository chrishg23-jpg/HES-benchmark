```python
# Act XIX: Mass-Coupling Interplay
import numpy as np
import matplotlib.pyplot as plt

# Parameters
N = 1000              # lattice size
timesteps = 300       # evolution steps
lambda_vals = np.linspace(0.5, 2.0, 20)  # sweep range
mass_vals = [0.1, 0.5, 1.0, 2.0, 5.0]    # test masses
seed = 42
np.random.seed(seed)

# Initial state generator
def init_state(N):
    return np.random.normal(0, 1, N)
```

```python
def evolve(state, timesteps, lambda_val, mass):
    history = []
    s = state.copy()
    for t in range(timesteps):
        # Update rule: mass-scaled attractor kernel
        s = s + (lambda_val / mass) * np.sin(s)
        history.append(s.copy())
    return np.array(history)
```

```python
def survival_metric(history):
    # Define survival as variance staying within stable bounds
    final_var = np.var(history[-1])
    return final_var
```

```python
results = {}

for m in mass_vals:
    survival_scores = []
    for lam in lambda_vals:
        state = init_state(N)
        hist = evolve(state, timesteps, lam, m)
        score = survival_metric(hist)
        survival_scores.append(score)
    results[m] = survival_scores
```

```python
plt.figure(figsize=(10,6))
for m in mass_vals:
    plt.plot(lambda_vals, results[m], label=f"mass={m}")
plt.xlabel("λ")
plt.ylabel("Survival Metric (Variance)")
plt.title("Mass-Coupling Interplay: Survival vs λ")
plt.legend()
plt.grid(True)
plt.show()
```
