```python
import numpy as np

# -------------------------------
# Step 1: Structure Formation Model
# -------------------------------
def structure_formation(delta, kappa, steps=100):
    """
    Simulate evolution of density fluctuations under entropic damping and curvature feedback.
    Returns fluctuation amplitude over time.
    """
    np.random.seed(42)
    fluctuations = np.random.normal(0, 1e-5, size=steps)  # initial perturbations
    growth = np.zeros(steps)
    growth[0] = fluctuations[0]

    for t in range(1, steps):
        curvature = np.sin(delta * t / steps) + np.cos(kappa * t / steps)
        damping = np.exp(-kappa * t / steps)
        feedback = 1 + delta * curvature / (1 + t)
        growth[t] = growth[t-1] * feedback * damping + fluctuations[t]

    return growth

# -------------------------------
# Step 2: Sweep δ and κ
# -------------------------------
delta_values = np.linspace(0.6, 0.8, 10)
kappa_values = np.linspace(0.6, 0.8, 10)

print("🪐 Act XXXVI: Entropic Structure Formation")
for delta in delta_values:
    for kappa in kappa_values:
        growth = structure_formation(delta, kappa)
        final_amp = np.abs(growth[-1])
        label = "✓" if final_amp > 1e-4 else "–"
        print(f"{label} δ = {delta:.4f}, κ = {kappa:.4f}, δ/κ = {delta/kappa:.3f} | Final Amplitude = {final_amp:.2e}")
```
