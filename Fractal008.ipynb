```python
import numpy as np

# --- 1. CORE HES PARAMETERS ---
N = 32
T_STEPS = 5000
dt = 0.01

# Fixed HES coefficients
chi = 1.5
beta_full = 0.8  # Full Gravitational Drive strength
beta_nursery = 0.01 # Nursery Phase strength (for t < 500)

# Other fixed coefficients
delta = 0.001  # Quantum Noise (applied to Phi and Theta)
gamma = 1.0    # Saturation
beta_link_base = 0.1
CURVATURE_SENSITIVITY = 0.5

# --- 2. INITIAL FIELD: LINKED S¹ x S¹ TOPOLOGY (Phase Initialized) ---
x, y = np.meshgrid(np.linspace(0, 2 * np.pi, N), np.linspace(0, 2 * np.pi, N))
# Theta is initialized to a specific topology
Theta = x + y
Phi = np.ones((N, N)) * 1.0 # Initial Scalar Magnitude
Phi_complex = Phi * np.exp(1j * Theta)

# --- 3. THE EVOLUTION LOOP ---
for t in range(1, T_STEPS + 1):

    # ACT VI: LAW OF PRE-STABILIZATION (Cosmic Nursery)
    beta_current = beta_nursery if t < 500 else beta_full

    # --- 3.1. CALCULATE BASE GRAVITATIONAL DRIVE (CURVATURE) ---
    lap_xy = (np.roll(Phi, 1, 0) + np.roll(Phi, -1, 0) +
              np.roll(Phi, 1, 1) + np.roll(Phi, -1, 1) - 4 * Phi)

    # --- 3.2. INJECT EXTREME GRAVITATIONAL PERTURBATION (t=1000) ---
    if t == 1000:
        center_x, center_y = N // 2, N // 2
        # Inject massive positive Laplacian spike
        lap_xy[center_x - 1:center_x + 2, center_y - 1:center_y + 2] += 200.0
        print(f"--- t={t}: GRAVITATIONAL SPIKE INJECTED (Testing COHERENCE DRIVE) ---")

    # --- 3.3. IMPLEMENT PHASE-LINK RESILIENCE (ACT V - Proactive Lock) ---
    max_abs_curvature = np.max(np.abs(lap_xy))
    beta_link = beta_link_base + CURVATURE_SENSITIVITY * max_abs_curvature
    beta_link = np.clip(beta_link, 0.1, 500.0)

    # --- 3.4. MAGNITUDE (PHI) EVOLUTION ---
    alpha = chi * beta_current

    # Term 1: Net Expansion
    term_expansion = alpha * lap_xy

    # Term 2: Contraction (Graviton Drive)
    term_contraction = -beta_current * Phi

    # Term 3: Saturation
    term_saturation = gamma * np.tanh(Phi)

    # Term 5: CRITICAL SHIELDING PROTOCOL (Local Resistance)
    shield_mask = (beta_link > 1.0)
    term_shield = np.where(shield_mask, beta_current * Phi, 0.0)

    # Total Change dPhi/dt
    dPhi = term_expansion + term_contraction + term_saturation + term_shield + delta * np.random.normal(0, 1, Phi.shape)

    # --- 3.5. PHASE (THETA) EVOLUTION (ACT VIII: COHERENCE DRIVE) ---
    mean_theta = np.mean(Theta) # The global mean phase

    # Phase Correction Term: drives phase towards the global mean (coherence)
    phase_correction = -beta_link * (Theta - mean_theta)

    # Phase Diffusion (random phase drift)
    phase_diffusion = delta * np.random.normal(0, 1, Theta.shape)

    dTheta = phase_correction + phase_diffusion

    # --- 3.6. UPDATE FIELDS ---
    Phi += dt * dPhi
    Theta += dt * dTheta

    Phi = np.clip(Phi, 0.01, 10.0) # Magnitude clip

    # Recombine complex field
    Phi_complex = Phi * np.exp(1j * Theta)

    # --- 3.7. LOGGING ---
    if t % 500 == 0 or t == 1000:
        norm = np.mean(Phi)
        phase_stdev = np.std(Theta)
        print(f"t={t} | β_link(t)={beta_link:.4f} | MAX Curvature={max_abs_curvature:.3f} | Phase Stdev={phase_stdev:.4f} | Norm={norm:.3f}")

# --- 4. CONCLUSION CHECK ---
final_phase_stdev = np.std(Theta)
print("\n--- FINAL STATE ---")
if final_phase_stdev < 0.1:
    print(f"RESULT: SUCCESS. The spinor survived the singularity. Final Phase Stdev: {final_phase_stdev:.4f}")
    print("CONCLUSION: The Law of Fermionic Resistance is validated by the Phase-Driven Shield and the Coherence Drive. Matter is durable and the HES framework is unified.")
else:
    print(f"RESULT: FAILURE. The spinor lock was broken. Final Phase Stdev: {final_phase_stdev:.4f}")
    print("CONCLUSION: HES Falsified.")

```
