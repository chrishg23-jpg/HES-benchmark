```python
# Chem014: Equilibrium & Partitioning via Curvature Symmetry

import numpy as np
import pandas as pd

# Constants
R = 8.314  # J/mol·K
T = 298    # Kelvin

# Define systems and curvature parameters
systems = [
    {
        "name": "Acid dissociation",
        "reaction": "HA ⇌ H⁺ + A⁻",
        "ΔG": -1.5,
        "interpretation": "Curvature liberation and entropic symmetry"
    },
    {
        "name": "CO₂ gas-liquid partitioning",
        "reaction": "CO₂(g) ⇌ CO₂(aq)",
        "ΔG": 1.2,
        "interpretation": "Curvature deepening in aqueous phase"
    },
    {
        "name": "Ligand binding",
        "reaction": "P + L ⇌ PL",
        "ΔG": 1.0,
        "interpretation": "Binding compresses curvature and entropy"
    },
    {
        "name": "Phase equilibrium",
        "reaction": "H₂O(l) ⇌ H₂O(g)",
        "ΔG": -1.3,
        "interpretation": "Entropic expansion in vapor phase"
    },
    {
        "name": "Ion exchange",
        "reaction": "Na⁺(aq) + K⁺(resin) ⇌ Na⁺(resin) + K⁺(aq)",
        "ΔG": -0.2,
        "interpretation": "Exchange symmetry with mild curvature shift"
    }
]

# Compute equilibrium properties
results = []
for s in systems:
    ΔG = s["ΔG"]
    K_eq = round(np.exp(-ΔG * 1000 / (R * T)), 3)
    partition = round(K_eq / (1 + K_eq), 3)
    results.append({
        "System": s["reaction"],
        "ΔG (kJ/mol)": ΔG,
        "K_eq": K_eq,
        "Partition Ratio / Binding Fraction": partition,
        "Curvature Interpretation": s["interpretation"]
    })

# Create and display dataframe
df = pd.DataFrame(results)
print(df)
```
