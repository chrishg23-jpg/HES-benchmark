```python
# Chem002: Valence & Bonding Simulation using HES attractor saturation

# Define atoms with valence electrons
atoms = {
    "H": {"Z": 1, "valence": 1},
    "O": {"Z": 8, "valence": 6},
    "N": {"Z": 7, "valence": 5},
    "Na": {"Z": 11, "valence": 1},
    "Cl": {"Z": 17, "valence": 7},
    "Mg": {"Z": 12, "valence": 2},
    "O2⁻": {"Z": 8, "valence": 8}  # oxide ion
}

# Define bonding rules
def simulate_bond(atom1, atom2):
    valence1 = atoms[atom1]["valence"]
    valence2 = atoms[atom2]["valence"]
    shared = min(valence1, valence2)
    entropic_gain = shared * 5  # arbitrary bonding benefit
    bond_type = "covalent" if shared >= 1 else "ionic"
    return {
        "pair": f"{atom1}-{atom2}",
        "bond_type": bond_type,
        "shared_electrons": shared,
        "entropic_gain": entropic_gain
    }

# Simulate bonds
pairs = [("H", "H"), ("O", "O"), ("N", "N"), ("Na", "Cl"), ("Mg", "O2⁻")]
results = [simulate_bond(a1, a2) for a1, a2 in pairs]

# Display results
for r in results:
    print(f"{r['pair']}: Type={r['bond_type']}, Shared={r['shared_electrons']}, Entropic Gain={r['entropic_gain']}")
```
