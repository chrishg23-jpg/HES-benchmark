```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import fft, fftfreq

# Replace these with your actual waveform arrays from ECO004 and ECO005
# Example placeholder arrays (use real data from your simulation)
waveform_shell = np.array([1800, 1600, 1400, 1200, 1000, 800, 600, 400, 200, 100, 50, 25, 10])
waveform_gaussian = np.array([1900, 1200, 300, 100, 50, 20, 10, 5])

# Pad both waveforms to equal length
length = max(len(waveform_shell), len(waveform_gaussian))
waveform_shell = np.pad(waveform_shell, (0, length - len(waveform_shell)), mode='constant')
waveform_gaussian = np.pad(waveform_gaussian, (0, length - len(waveform_gaussian)), mode='constant')

# Sampling interval (1 timestep)
dt = 1.0
freqs = fftfreq(length, dt)[:length // 2]

# Compute FFT
fft_shell = fft(waveform_shell)
fft_gaussian = fft(waveform_gaussian)

# Amplitude spectra
amp_shell = np.abs(fft_shell[:length // 2])
amp_gaussian = np.abs(fft_gaussian[:length // 2])

# Plot comparison
plt.figure(figsize=(12, 6))
plt.plot(freqs, amp_shell, label='Shell Field Spectrum', color='purple')
plt.plot(freqs, amp_gaussian, label='Gaussian Field Spectrum', color='orange')
plt.title("Fourier Spectra of Collapse Echoes")
plt.xlabel("Frequency")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
