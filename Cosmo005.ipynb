```python
import numpy as np

# -------------------------------
# Step 1: Reheating Model
# -------------------------------
def entropic_reheating(delta, kappa, steps=100):
    """
    Simulate energy redistribution after structure formation.
    Returns energy profile and reheating metrics.
    """
    np.random.seed(42)
    energy = np.zeros(steps)
    residual = 0.0

    for t in range(steps):
        curvature = np.sin(delta * t / steps) + np.cos(kappa * t / steps)
        damping = np.exp(-kappa * t / steps)
        injection = np.abs(curvature * delta / (1 + t))
        residual += injection * damping
        energy[t] = residual

    reheating_rate = energy[10] / energy[0] if energy[0] != 0 else 0
    final_energy = energy[-1]
    return energy, reheating_rate, final_energy

# -------------------------------
# Step 2: Sweep δ and κ
# -------------------------------
delta_values = np.linspace(0.6, 0.8, 10)
kappa_values = np.linspace(0.6, 0.8, 10)

print("🪐 Act XXXVII: Entropic Reheating")
for delta in delta_values:
    for kappa in kappa_values:
        energy, rate, final = entropic_reheating(delta, kappa)
        label = "✓" if rate > 2.0 else "–"
        print(f"{label} δ = {delta:.4f}, κ = {kappa:.4f}, δ/κ = {delta/kappa:.3f} | Reheating Rate = {rate:.2f}, Final Energy = {final:.4f}")
```
