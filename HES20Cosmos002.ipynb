```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter1d
```

```python
k_vals = np.arange(1, 101)
t_final = 2000
k_dom = 50

A_k = np.exp(-np.abs(k_vals - k_dom) / 10) * np.tanh(t_final / 500)
```

```python
P_k = A_k**2
P_k /= np.max(P_k)  # Normalize to peak at 1
```

```python
P_k_smooth = gaussian_filter1d(P_k, sigma=2)
noise = np.random.normal(0, 0.05, size=len(P_k))
P_k_obs = P_k_smooth + noise
```

```python
plt.figure(figsize=(10, 6))
plt.plot(k_vals, P_k, label='HES Modal Memory', color='blue')
plt.plot(k_vals, P_k_obs, label='Synthetic Galaxy Clustering', color='red', linestyle='--')
plt.xlabel('Mode k')
plt.ylabel('Power Spectrum P(k)')
plt.title('Act LV: Modal Memory in Galaxy Clustering')
plt.legend()
plt.grid(True)
plt.show()
```
