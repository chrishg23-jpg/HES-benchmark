```python
# Chem004: Hybridization & Resonance from HES curvature interference

# Define molecules with hybridization and resonance info
molecules = {
    "CH4":   {"hybridization": "sp3", "geometry": "tetrahedral", "resonance": 0, "base_gain": 15},
    "C2H4":  {"hybridization": "sp2", "geometry": "trigonal planar", "resonance": 0, "base_gain": 10},
    "C2H2":  {"hybridization": "sp",  "geometry": "linear", "resonance": 0, "base_gain": 5},
    "O3":    {"hybridization": "sp2", "geometry": "bent", "resonance": 2, "base_gain": 10},
    "NO2":   {"hybridization": "sp2", "geometry": "bent", "resonance": 2, "base_gain": 10},
    "C6H6":  {"hybridization": "sp2", "geometry": "planar hexagon", "resonance": 6, "base_gain": 10}
}

# Calculate entropic gain from hybridization and resonance
def calculate_entropic_gain(mol):
    resonance_bonus = mol["resonance"] * 2  # each resonance structure adds 2 units
    total_gain = mol["base_gain"] + resonance_bonus
    return total_gain

# Run simulation
print("═" * 70)
print("Chem004: Hybridization & Resonance Simulation")
print("═" * 70)
for name, mol in molecules.items():
    gain = calculate_entropic_gain(mol)
    print(f"{name:5} | Hybridization: {mol['hybridization']:4} | Geometry: {mol['geometry']:18} | "
          f"Resonance: {mol['resonance']} | Entropic Gain: {gain}")

# Optional: visualize entropic gain vs resonance count
import matplotlib.pyplot as plt

names = list(molecules.keys())
resonance_counts = [molecules[m]["resonance"] for m in names]
gains = [calculate_entropic_gain(molecules[m]) for m in names]
colors = ['blue' if r == 0 else 'purple' for r in resonance_counts]

plt.figure(figsize=(10, 5))
plt.bar(names, gains, color=colors)
plt.title("Entropic Gain from Hybridization and Resonance")
plt.xlabel("Molecule")
plt.ylabel("Entropic Gain")
plt.grid(True, axis='y', alpha=0.3)
plt.show()
```
